```python
import pandas as pd
import numpy as np

def generate_data(n=200):
    days = np.arange(1, n+1)
    travel_km = np.random.uniform(2, 50, n)
    electricity = np.random.uniform(1, 15, n)
    fuel = np.random.uniform(0.2, 5, n)

    co2 = 0.21*travel_km + 0.82*electricity + 2.31*fuel + np.random.normal(0, 2, n)

    df = pd.DataFrame({
        "day": days,
        "travel_km": travel_km,
        "electricity": electricity,
        "fuel": fuel,
        "co2": co2
    })

    df.to_csv("emission_history.csv", index=False)
    print("Dataset generated: emission_history.csv")

if __name__ == "__main__":
    generate_data()
```
